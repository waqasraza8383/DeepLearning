```python
from sklearn.datasets import make_classification
import numpy as np
X,y=make_classification(n_classes=2,n_samples=100,n_features=2,random_state=41,n_clusters_per_class=1,n_informative=1,n_redundant=0,class_sep=10,hypercube=False)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def perceptron(X,y):
    X=np.insert(X,0,1,axis=1)
    weights=np.ones(X.shape[1])
    lr=0.1
    for i in range(100):
        j=np.random.randint(0,100)
        y_hat=step(np.dot(X[j],weights))
        weights=weights+lr*(y[j]-y_hat)*X[j]

    #weights 0 mean intercept and  weights[1:]  coefficent
    return weights[0],weights[1:]
    
```

```python
def step(z):
    return 1 if z>0 else 0
```

```python
intercept_,coef=perceptron(X,y)
```
